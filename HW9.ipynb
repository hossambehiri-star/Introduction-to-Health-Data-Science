### Q14. How do you create a cross-tabulation table that shows the frequency count of `Department` (rows) and `Job Title` (columns) in a DataFrame `df`?

```python
# Generic example using the heart_failure dataset
pd.crosstab(df['diabetes'], df['anaemia'])
```

### Q15. How do you create a cross-tabulation table that shows the mean `Salary` for each combination of `Department` (rows) and `Job Title` (columns) in a DataFrame `df`?

```python
# Generic example using the heart_failure dataset
pd.crosstab(df['diabetes'], df['anaemia'], values=df['age'], aggfunc='mean')
```
