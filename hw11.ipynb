# HW11 - Confidence Intervals & Bootstrapping

```python
import numpy as np
from scipy import stats

# Seed the pseudo-random number generator
numpy.random.seed(28)

# Create an array of random values from a normal distribution with a mean of 85 and a standard deviation of 10
dBP = numpy.random.normal(85, 10, 100000)
```

### Q1. Select a random sample of 40 individuals from the population and assign the created numpy array to the variable `sample`.

```python
numpy.random.seed(12)
sample = numpy.random.choice(dBP, size=40, replace=False)
sample
```

### Q2. Calculate the sample mean and assign the value to the variable `sample_mean`.

```python
sample_mean = sample.mean()
sample_mean
```

### Q3. Calculate the standard deviation of the sample and assign the value to the variable `sample_std`.

```python
sample_std = sample.std(ddof=1)
sample_std
```

### Q4. Calculate the standard error of the sample and assign the value to the variable `standard_error`.

```python
standard_error = sample_std / numpy.sqrt(len(sample))
standard_error
```

### Q5. Calculate the confidence coefficient for a 95% confidence level, given the sample size and degrees of freedom, and assign the value to the variable `c`.

```python
degrees_of_freedom = len(sample) - 1
c = stats.t.ppf(0.975, df=degrees_of_freedom)
c
```

### Q6. Calculate the lower bound of the confidence interval for the sample mean and assign the result to the variable `lower_bound`.

```python
lower_bound = sample_mean - c * standard_error
lower_bound
```

### Q7. Calculate the upper bound of the confidence interval for the sample mean and assign the result to the variable `upper_bound`.

```python
upper_bound = sample_mean + c * standard_error
upper_bound
```

### Q8. Create an array of 1000 bootstrapped sample means (using the sample size of 40) and assign the array to the variable `bootstrapped_means`, sorted in ascending order.

```python
numpy.random.seed(12)
bootstrapped_means = numpy.array([
    numpy.random.choice(sample, size=40, replace=True).mean()
    for _ in range(1000)
])
bootstrapped_means = numpy.sort(bootstrapped_means)
bootstrapped_means
```

### Q9. Show the bootstrapped value representing the 2.5th percentile.

```python
numpy.percentile(bootstrapped_means, 2.5)
```

### Q10. Show the value bootstrapped representing the 97.5th percentile.

```python
numpy.percentile(bootstrapped_means, 97.5)
```
