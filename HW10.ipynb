# Chapter 9 - Seaborn Visualization Questions

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('https://raw.githubusercontent.com/juanklopper/TutorialData/main/heart_failure.csv')
```

### Q1. How do you import the Seaborn package in Python?

```python
import seaborn as sns
```

### Q2. How do you create a scatter plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.scatterplot(x='age', y='serum_creatinine', data=df)
plt.show()
```

### Q3. How do you create a line plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.lineplot(x='age', y='serum_creatinine', data=df)
plt.show()
```

### Q4. How do you create a histogram in Seaborn for a variable `x` in a DataFrame `df`?

```python
ax = sns.histplot(x='age', data=df)
plt.show()
```

### Q5. How do you create a stacked histogram in Seaborn for variables `x` and hue `y` in a DataFrame `df`?

```python
ax = sns.histplot(x='age', hue='death', data=df, multiple='stack')
plt.show()
```

### Q6. How do you create a heatmap in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
# Heatmap is typically used on a correlation matrix or pivot table
corr_matrix = df[['age', 'serum_creatinine', 'serum_sodium', 'ejection_fraction', 'platelets']].corr()
ax = sns.heatmap(corr_matrix, annot=True)
plt.show()
```

### Q7. How do you create a box plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.boxplot(x='death', y='age', data=df)
plt.show()
```

### Q8. How do you create a violin plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.violinplot(x='death', y='age', data=df)
plt.show()
```

### Q9. How do you create a bar plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.barplot(x='death', y='age', data=df)
plt.show()
```

### Q10. How do you add a title to a Seaborn plot assigned to the variable `ax` with the title 'My Title'?

```python
ax = sns.boxplot(x='death', y='age', data=df)
ax.set_title('My Title')
plt.show()
```

### Q11. How do you save a Seaborn plot `ax` to a file named 'output.png'?

```python
ax = sns.boxplot(x='death', y='age', data=df)
ax.get_figure().savefig('output.png')
plt.show()
```

### Q12. How do you set the `x` and `y` axis labels to 'X Label' and 'Y Label' respectively for a Seaborn plot `ax`?

```python
ax = sns.scatterplot(x='age', y='serum_creatinine', data=df)
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
plt.show()
```

### Q13. How do you change the color palette of a Seaborn plot to 'Blues'?

```python
ax = sns.boxplot(x='death', y='age', data=df, palette='Blues')
plt.show()
```

### Q14. How do you display a Seaborn plot with a logarithmic scale on the y-axis?

```python
ax = sns.scatterplot(x='age', y='creatinine_phosphokinase', data=df)
ax.set_yscale('log')
plt.show()
```

### Q15. How do you add a regression line to a scatter plot in Seaborn for variables `x` and `y` in a DataFrame `df`?

```python
ax = sns.regplot(x='age', y='serum_creatinine', data=df)
plt.show()
```
